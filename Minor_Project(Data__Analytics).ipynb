# 9. Customer Segmentation

Customer segmentation is the process of dividing customers into different groups based on similarities in their characteristics and purchasing behaviour. In this project, K-Means clustering is used to identify customer segments based on Age, Purchase Amount, Review Rating, and Previous Purchases.

## 9.1 Feature Selection

The numerical features used for customer segmentation are selected from the dataset. These features represent customer demographics, spending behaviour, satisfaction level, and purchase frequency.

```python
segmentation_data = data[
    ["Age", "Purchase Amount (USD)", "Review Rating", "Previous Purchases"]
]

segmentation_data.head()
```

### Observation

Four numerical features were selected for customer segmentation: Age, Purchase Amount, Review Rating, and Previous Purchases.

## 9.2 Data Standardization

Since the selected features have different ranges, standardization is performed to bring all features into a similar scale. This prevents any single feature from dominating the clustering process.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

segmentation_scaled = scaler.fit_transform(segmentation_data)

segmentation_scaled[:5]
```

### Observation

The numerical features were successfully standardized and prepared for the clustering algorithm.

## 9.3 Elbow Method

The Elbow Method is used to determine the optimal number of clusters for the K-Means algorithm. It calculates the within-cluster variation for different values of K and helps identify the point where adding more clusters provides less improvement.

```python
from sklearn.cluster import KMeans

inertia = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(segmentation_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8,5))

plt.plot(range(1,11), inertia, marker="o")

plt.title("Elbow Method for Optimal Number of Clusters")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")

plt.grid(alpha=0.3)
plt.show()
```

### Observation

The Elbow Method was used to analyse different cluster values. Based on the elbow point in the graph, **3 clusters** were selected for customer segmentation.

## 9.4 K-Means Clustering

K-Means clustering is applied to group customers with similar characteristics into different segments.

```python
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

data["Cluster"] = kmeans.fit_predict(segmentation_scaled)

data[["Age", "Purchase Amount (USD)", "Cluster"]].head()
```

### Observation

The customers were successfully divided into three different clusters using the K-Means clustering algorithm.

## 9.5 Cluster Distribution

The number of customers present in each cluster is analysed to understand the size of each customer group.

```python
data["Cluster"].value_counts()
```

### Observation

The cluster distribution shows the number of customers belonging to each identified customer segment.

## 9.6 Cluster Characteristics

The average values of each feature are calculated for every cluster to understand the characteristics and behaviour of different customer segments.

```python
cluster_summary = data.groupby("Cluster")[[
    "Age",
    "Purchase Amount (USD)",
    "Review Rating",
    "Previous Purchases"
]].mean()

cluster_summary
```

### Observation

The cluster summary shows the average characteristics of each customer segment. Each cluster represents customers with different purchasing behaviours based on age, spending amount, ratings, and previous purchases.

## 9.7 Customer Segmentation Visualization

A scatter plot is created to visualize the distribution of customers across different clusters based on Age and Purchase Amount.

```python
plt.figure(figsize=(8,6))

plt.scatter(
    data["Age"],
    data["Purchase Amount (USD)"],
    c=data["Cluster"],
    cmap="viridis"
)

plt.title("Customer Segmentation using K-Means")
plt.xlabel("Age")
plt.ylabel("Purchase Amount (USD)")

plt.grid(alpha=0.3)

plt.show()
```

### Observation

The scatter plot shows the distribution of customers into three different clusters. Customers with similar age and spending patterns are grouped together, helping identify different customer segments.
